## Feature Engineerig and Visualization

There are some feature transformation to do in order train a model:
1. Change the target values from the precise result to 1, X, 2
2. Choose one between Ball possession_Casa and Ball possession_Fuori
3. Extract from Accurate passes_Casa and Accurate passes_Fuori the percentage Es. "200 (64%)" --> "64" 
4. Visualize Red cards_Casa and Red cards_Fuori in order to assess if there are too many values "0"

5. Other things to do:
    - Map the team names to their IDs (or simply change the categorical values to numerical using an encoding)
    - Remove the now useless features: Gol Casa and Gol Fuori
    - Calculate the mean of all xGs and shots for each team

### 1. Change the target value from precise to 1, X, 2

```python
import numpy as np

conditions = [
    df_first['Gol Casa'] > df_first['Gol Fuori'],
    df_first['Gol Casa'] == df_first['Gol Fuori'],
    df_first['Gol Casa'] < df_first['Gol Fuori']
]
# map[1] sta a conditions[1]
map = ['1', 'X', '2']

#Sovrascrivo 'Risultato' con i nuovi valori
df_first.loc[:, 'Risultato'] = np.select(conditions, map, default='Errore')

print(df_first[['Squadra Casa', 'Squadra Fuori', 'Gol Casa', 'Gol Fuori', 'Risultato']].head())
```

#### Visualize match result distribution

```python
conteggi = df_first['Risultato'].value_counts()

# 2. Crea la figura e il grafico a torta
plt.figure(figsize=(8, 6))
plt.pie(conteggi, 
        labels=conteggi.index,
        autopct='%1.1f%%',
        startangle=90,
        colors=['#ff5555', '#55ff55', '#5555ff'],
        wedgeprops={'edgecolor': 'black'})
plt.title('Match Result Distribution (1, X, 2)')
plt.axis('equal')
plt.show()
```

### 3. Extract the percentage from Accurate passes_Casa and Accurate passes_Fuori

```python
df_second = df.copy()

colonne_passaggi = ['Accurate passes_Casa', 'Accurate passes_Fuori']

for col in colonne_passaggi:

    df_second[col] = pd.to_numeric(
        df_second[col].str.extract(r'\((\d+)%\)', expand=False), # \( cerca una tonda, (\d+) cerca una cifra o piu e la estrae se è seguita da una percentuale %\)
        errors='coerce'
    )

print(df_second[['Squadra Casa', 'Squadra Fuori', 'Accurate passes_Casa', 'Accurate passes_Fuori']].head())
```

### 5. Other Corrections

#### a. Map the team names to their IDs (or simply change the categorical values to numerical using an encoding)


```python
team_map = {
    'Atalanta': 8524,
    'Bologna': 9857,
    'Cagliari': 8529,
    'Como': 10171,
    'Cremonese': 7801,
    'Empoli': 8534,
    'Fiorentina': 8535,
    'Frosinone': 9891,
    'Genoa': 10233,
    'Hellas Verona': 9876,
    'Inter': 8636,
    'Juventus': 9885,
    'Lazio': 8543,
    'Lecce': 9888,
    'Milan': 8564,
    'Monza': 6504,
    'Napoli': 9875,
    'Parma': 10167,
    'Pisa': 6479,
    'Roma': 8686,
    'Salernitana': 6480,
    'Sassuolo': 7943,
    'Torino': 9804,
    'Udinese': 8600,
    'Venezia': 7881
}

df_second['Home_ID'] = df_second['Squadra Casa'].map(team_map)
df_second['Away_ID'] = df_second['Squadra Fuori'].map(team_map)

print(df_second[['Squadra Casa', 'Home_ID', 'Squadra Fuori', 'Away_ID']].head())
```

#### b. Remove the now useless features

- Red Cards are now considered useless because of their distributions. They are zero for the vast mojority of the entries
- Squadra Casa and Squadra Fuori both need to be removed
- Gol Casa and Gol Fuori are useless given "Risultato"

```python
df_third = df_second.drop(columns=["Red cards_Casa", "Red cards_Fuori", "Squadra Casa", "Squadra Fuori", "Gol Casa", "Gol Fuori"]).copy()
df_third.info()
```

#### c. Calculate the mean of all xGs and shots for each team (both when the team is Home and when it is Away)

```python
# 2. Definiamo le feature da mediare, divise tra Casa e Trasferta
colonne_casa = [
    'Expected goals (xG)_Casa', 
    'Shots inside box_Casa', 
    'Shots outside box_Casa',
    'xG open play_Casa', 
    'xG set play_Casa', 
    'xG on target (xGOT)_Casa'
]

colonne_fuori = [
    'Expected goals (xG)_Fuori', 
    'Shots inside box_Fuori', 
    'Shots outside box_Fuori',
    'xG open play_Fuori', 
    'xG set play_Fuori', 
    'xG on target (xGOT)_Fuori'
]


# group by Squadra Casa and apply mean over the home features
medie_casa = df_second.groupby('Squadra Casa')[colonne_casa].mean()

# group by Squadra Fuori and apply mean over the away features
medie_fuori = df_second.groupby('Squadra Fuori')[colonne_fuori].mean()



df_medie_squadre = medie_casa.join(medie_fuori, how='inner')

# Aggiungo ID
df_medie_squadre['Team_ID'] = df_medie_squadre.index.map(team_map)

# ID come prima colonna
colonne_riordinate = ['Team_ID'] + [col for col in df_medie_squadre.columns if col != 'Team_ID']
df_medie_squadre = df_medie_squadre[colonne_riordinate]
```

#### d. Team Rating Visualization through each Match
A function that allows us to see, for the team that we give as an input, the trend of the team Rating when it plays Home and Away

```python
team_map = {
    'Atalanta': 8524, 'Bologna': 9857, 'Cagliari': 8529, 'Como': 10171,
    'Cremonese': 7801, 'Empoli': 8534, 'Fiorentina': 8535, 'Frosinone': 9891,
    'Genoa': 10233, 'Hellas Verona': 9876, 'Inter': 8636, 'Juventus': 9885,
    'Lazio': 8543, 'Lecce': 9888, 'Milan': 8564, 'Monza': 6504,
    'Napoli': 9875, 'Parma': 10167, 'Pisa': 6479, 'Roma': 8686,
    'Salernitana': 6480, 'Sassuolo': 7943, 'Torino': 9804, 'Udinese': 8600,
    'Venezia': 7881
}
```

```python
def plot_team_rating_trend(team_name):
   
    team_id = team_map.get(team_name)
    if team_id is None:
        print(f"Errore: La squadra '{team_name}' non è presente nel dizionario. Controlla l'ortografia.")
        return

    
    home_matches = df_third[df_third['Home_ID'] == team_id].copy()
    away_matches = df_third[df_third['Away_ID'] == team_id].copy()

    # due df
    home_data = pd.DataFrame({
        'Timeline': home_matches.index,
        'Rating': home_matches['HomeTeamRating'],
        'Venue': 'Home'
    })
    away_data = pd.DataFrame({
        'Timeline': away_matches.index,
        'Rating': away_matches['AwayTeamRating'],
        'Venue': 'Away'
    })


    team_df = pd.concat([home_data, away_data]).sort_values('Timeline').reset_index(drop=True)
    if team_df.empty:
        print(f"Nessuna partita trovata per {team_name} nel dataset.")
        return
        
    team_df['Match_Number'] = team_df.index + 1

    
    plt.figure(figsize=(12, 6))
    
    sns.lineplot(data=team_df, 
                 x='Match_Number', 
                 y='Rating', 
                 hue='Venue', 
                 marker='o',
                 palette={'Home': '#4CB391', 'Away': '#E66C37'}, 
                 linewidth=2.5,
                 markersize=8)

    plt.title(f'Rating Trend for {team_name}: Home vs Away', fontsize=16, fontweight='bold', pad=15)
    plt.xlabel('Match Number (Chronological)', fontsize=12)
    plt.ylabel('Team Rating', fontsize=12)
    
    even_ticks = [x for x in team_df['Match_Number'] if x % 2 == 0]
    plt.xticks(even_ticks, rotation=70) 
    
    plt.grid(alpha=0.3)
    plt.legend(title='Venue', loc='upper left')
    plt.tight_layout()
    plt.show()
```

```python
plot_team_rating_trend('Como')
```

```python
plot_team_rating_trend("Salernitana")
```
